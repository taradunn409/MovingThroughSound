```python
# Hello!! Welcome to my Creative Coding Project A2
# Moving Through Sound - Tara Dunn 25805767

# I will start by installing the code libraries needed to achieve my coding prototype.
# I am using the command %pip instead of !pip becuase %pip installs Python packages directly inside the active Jupyter kernel I am coding within
# This ensures that the libraries I intall end up where my notebook can find them
# Here I am asking it to install mediapipe (hand tracking) and open cv/cv2 (webcam and video window)
%pip install mediapipe opencv-python 

# Pygame-ce a community version of pygame that is used to develop video games and multimedia applications
# I am using its mixer module, which loads sound files, plays several at once, and lets me change each one's volume while it plays
# I chose pygame-ce instead of the original pygame because installing the original failed on my computer, pygame-ce is better for newer Python versions
# This is important for later when I want to add sounds to my handlandmarks
%pip install pygame-ce

# Once installed, I have imported these dependancies into my coding environment
# I have also written "import mediapipe as mp" to create a shortcut, which saves me typing out MediaPipe in every line and keeps my code cleaner
import mediapipe as mp
import cv2
import pygame

# Firstly, I will set up everything MediaPipe needs for hand tracking such as the Tasks module imports, the drawing tools, the model file and the detector options
# My YouTube tutorial used MediaPipe's older "solutions" methods but it has now moved away from this and toward a modular framework called Tasks
# So I have imported the modules from the new framework 
# BaseOptions is part of the Tasks framework that tells mp where my model file is
from mediapipe.tasks.python import BaseOptions
# And this is the vision sub module that hold the vision tools such as the hand detector, hand landmarker options and the running mode
from mediapipe.tasks.python import vision

# mp.tasks.vision.drawing_utils are the mp built in drawing tools which I will use later on in the code to draw lines between my hand landmarks
# I have named the variable mp_drawing as a shortcut and stored the drawing utilities under it
mp_drawing = mp.tasks.vision.drawing_utils

# Here I have created a variable that stores the location of the task file "hand_landmarker.task", so I can use it as a shortcut in the next parts of code
model_path = "hand_landmarker.task"

# Now I am changing the options of the detector's settings to suit my project's goal
# To do this I have created a variable that holds the values of each setting I am wishing to change inside the brackets
# This code is all available on the MediaPipe Tasks website under "Hand landmarks detection guide for Python"
options = vision.HandLandmarkerOptions(
    
    # Here I am using the model path variable I created before to load the hand landmarker task file
    base_options=BaseOptions(model_asset_path=model_path),
    
    # The video option I have chosen to use from the model gives mp one frame from the video capture at a time, in order and returns the result
    running_mode=vision.RunningMode.VIDEO,
    
    # This tells the code the number of hands to look for
    # I have chosen one for simplicity and more control for the user over the sounds that are played
    num_hands=1,
    
    # This code decides how sure mp is that the object on screen is a hand from a scale of 0 to 1 
    # I have chosen 0.8 so mp is pretty strict and prevents it from mistaking other objects as a hand
    min_hand_detection_confidence=0.8,
    
    # And finally this setting decides how sure mp must be to continue tracking the hand that it has found also on a scale of 0 to 1
    # I have chosen 0.5 so the detection doesn't drop out too easily, especially when tracking faster hand movements
    min_tracking_confidence=0.5
)

# Then, I need to set up my sound system
# This piece of code starts the sound system, it first has to be inisitialised before it can play anything
pygame.mixer.init()

# My sound files were not in my notebook's folder at first, just like the hand landmarker model, so python could not find them
# I had to open Jupyter's file browser where I could then add in my files
# I created a folder called Sound_Effects and uploaded my three .mp3 files inside it

# pygame.mixer.Sound loads each file into memory and gives me a sound object I can control
# The paths include the folder name (which I created), so "Sound_Effects/rain.mp3" means "go into the Sound_Effects folder, then find rain.mp3"
# The names must match my files exactly, including capital letters and the .mp3 ending so it can be found
# I have then created variables to store my sounds under to refer back to later in the code
rain = pygame.mixer.Sound("Sound_Effects/rain.mp3")
wind = pygame.mixer.Sound("Sound_Effects/wind_chimes.mp3")
birds = pygame.mixer.Sound("Sound_Effects/forest_birds.mp3")

# all_sounds is a tuple that groups my three sounds, so I can loop through them together later.
all_sounds = (rain, wind, birds)

# Then I need to define the function "volume_for" which is setting up the volume parameters within the screen for later
# x is my hand position from 0 (left edge) to 1 (right edge) and centre is the screen position where this sound should be at its loudest.
def volume_for(x, centre, width):

# The volume of each sound is dependant of where my index fingertip lies on the screen, so I need to calculate the plane where each sound will be the loudest
# I am using abs() which is python's absolute value function to get the distance from my hand to the sound's centre, keeping it in the positive ranges
# I am then dividing that value by 0.5 to get the zone width for each sound so 0-0.5 sound 1, 0.25-0.75 sound 2 and 0.5-1 sound 3
# Then subtracting it from 1 flips it, so the volume is 1 (full) when my hand is right on the centre and falls to 0 half a screen away.\
    volume = 1 - abs(x - centre) / width

    # max(volume, 0) stops the volume going negative, so the lowest it can be is 0 (silent).
    # return hands the finished number back to the line that called the function.
    return max(volume, 0)
```

```python
# Now that I have finished installing and importing my dependancies, I can move onto the next part of my project
# I have broken up my code into separate cells to create cleaner production line where each step finishes its job entirely before the next step begins
# IT also makes my code easier to debug if I make mistakes because I can test each part of my project individually for problems

# I have added these next two lines of code before the video capture so that both of their settings are defined and exist when the loop reaches them

# I have been debating whether to have the landmarks of the hand visable on the webcam for aesthetic purposes
# Seeing the drawn points is probably the easier way to see if the tracking is working properly and it is a lot easier than reading the array of printed numbers if I printed the results list
# It will also let me see if there is a problem with the sound or tracking code when I add the sounds later
# I think that If I was doing an art exhibition I wouldn't want them displayed but whilst coding it is nice to visualise
# So, I have added and On/Off function that I can use when I do or don't want to see the landmarker drawings
SHOW_HAND = True

# Then I am creating a variable which holds my hand landmark connections and will eventaully be needed to draw the lines between hand landmarks
# HAND_CONNECTIONS is mp's built in list of which landmarks are joined by a line on the hand
# I have had to change it from the old "solutions" method which was mp_hands.HAND_CONNECTIONS, to suit the new tasks framework 
# I found this class variable under the vision module on the mp website
HAND_CONNECTIONS = mp.tasks.vision.HandLandmarksConnections.HAND_CONNECTIONS

# Lastly, before I open up the webcam, I need to set up the volume and sound settings for my hand tracker
# FADE_STEP is how much the master volume changes on every frame when a hand appears or disappears
# I need this to smoothen out the blend of my sound effects as they change from one another, such as rain fading out as the birds fade in, with no sound starting or stopping
# 0.03 is just a starting value but I can always change the value until the fading feels right
FADE_STEP = 0.03

# Master is an overall volume from 0 (silent) to 1 (full volume) that multiplies every sound's volume
# I am placing htis outside of the loop so that the volume starts at 0 each time the code runs
# This means that the space will be silent until someone puts a hand in front of the camera
master = 0.0

# This variable will represent the detected hand, more specifically the last location it was detected across the screen from 0 (left edge of the screen) to 1 (the right edge)
# This is important becuase my volume lined run on every frame, including frames where no hand is found, so x must always exist or python gives an error
# So when my hand leaves the screen, x should keep its last value and the sound fades out nstead of stopping
# I am using the value 0.5 or the middle of the screen as a place holder, but the position isn't esstential anyways since the master volume starts at 0 and the sound effects wont begin until a hand is detected
x = 0.5

# Here I am creating a little loop that goes through my three sounds all at once, and the variable "sound" is the current line of loop it is running
# This means that the two indented lines below run three times, once for rain, once for wind chimes and once for birds.
for sound in all_sounds:
    
    # In the loop I am making every sound start silent, because the master volume starts at 0
    sound.set_volume(0)
    
    # Then I am starting the sound with play and making it repeat it forever through (loops=-1)
    # The "-1" value in this loop causes all three sounds to be always playing and only their volumes change, making the blend between sounds smooth
    sound.play(loops=-1)
    
# Now I will open the webcam and set up the video display

# cv2.VideoCapture(0) opens up the webcam and display the image in a sepreate window
# I am storing it in the variable cap (capture) as a shortcut for later
# The number "0" in cv2.VideoCapture(0) chooses which camera to use, as my laptop only has one webcam, this number defaults to zero
cap = cv2.VideoCapture(0)

# The old "solutions" method from mp kept track of frames by itself but since its shift, it now needs to be done manually
# I have created a frame counter as the video variation of the task construction needs a timestamp that goes up with each frame to read them in order 
frame_count=0

# This is where the detector is actually set up
# I am using a with statement as a context manager which ensures that it sets the detector up "vision.HandLandmarker.create_from_options(options)", runs the code below it then closes the detector after its done
# I have added "as landmarker" to give the detector a name I can refer to in the code below as a shortcut
with vision.HandLandmarker.create_from_options(options) as landmarker:
    
    # In this section I am using Boolean Logic to create a loop, so as long as the webcam connection is active and open, my video feed will be read and the code will run
    while cap.isOpened():

        # These next sections of code are all placed before cv2.imshow as they alter the displayed image
        # If it were after cv2.imshow the changes would not show and the code would not run properly
        
        # cap.read() grabs a frame and returns the values of ret and frame which I have named here as variables
        # "ret" is a boolean true/false, so if the "frame" contains image data then the camera feed from the webcam can be read correctly
        ret, frame = cap.read()

        # If ret is false, the camera has failed to give a frame and the variable frame would be empty, which would crash cv2.imshow
        # So before that happens, I have added, if not ret (if false), to break the loop
        if not ret:
            break

        # I have added image flip to my code as an extra, so the image that I see mirrors what I am doing in real life
        # This not only makes more sense in my brain but it coorelates to the sound values I will add later on in my code
        # The variable of the flipped frame is now image, so now when referring to a frame in the later code, it has changed to this altered image
        image = cv2.flip(frame, 1)

        # Instead of changing the colour of the real image, I am making a copy of the frame under the variable rgb to send to mp to read
        # The default image colour order for cv2 is BGR but mp needs the RGB format, so here I am swapping the order of the colour channels in the copy
        # I am passing through the image and specifying what colour conversion I want cv2.COLOR_BGR2RGB, so  "BGR" to "RGB"
        # This then returns a new picture and leave the original “image” untouched so I can draw my detections on the original BGR/cv2 version
        # I have named my variable as rgb as I have already used "image" and "frame as variables
        rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

        # Now, I have to prepare the data for mp to read, using the mp website I have found the format that mp uses to read data and adapted it to my code and variables
        # The code mp.Image wraps my copied rgb picture in the container type that mp’s detector requires, because it will not accept a plain block of numbers
        # It is taking my copied rgb picture, putting it in a mp image container, labeling it as rgb, and storing it in mp_Image
        mp_image = mp.Image(image_format=mp.ImageFormat.SRGB, data=rgb)

        # Then to continue sending frames to mp I have to increase the frame count as the loop continues 
        # The frame_count was set to zero before the loop, so "+= 1" means add one to the current value each time the loop starts again, so each time the loop runs, the frame count increases
        frame_count += 1

        # I am creating the varaible fps which I am using to calculate the frame count's timestamps for mp
        # cap.get() is a method that retrieves data from the camera, which in this case is cv2.CAP_PROP_FPS, cv2's name for fps
        # So I am asking cv2 to retrieve the frames per second and storing it inside fps
        fps = cap.get(cv2.CAP_PROP_FPS)

        # This part here is a little fail safe if my webcam doesn't report its frame rate or gives back a non usable number
        # To make sure the webcam doesn'y crash, I am using a boolean logic statement to check if fps is less than or equal to 0
        # 30 fps is a pretty basic rendering of frames for a computer webcam, so if the camera doesn't give me a usable number, then I have set the fps to equal 30 as a fallback
        if fps <= 0:
            fps = 30

        # Retrieving the fps is important becuase video variation mode of mp's task needs a timestamp in milliseconds that always increases
        # I am using 1000 becuase there are 1000 milliseconds per second and dividing it by the fps to find how long one frame lasts
        # Then I am multiplying it by the frame_count to give the time of the current frame since the program started running
        # I have also added "int" in front of the brackets to convert it to a interger which mp can read
        frame_timestamp_ms = int(1000 * frame_count / fps)

        # This is the hand detector itself, I have given it the wrapped image and the timestamp which will then return the result to this variable
        hand_landmarker_result = landmarker.detect_for_video(mp_image, frame_timestamp_ms)

        # hand_landmarker_result.hand_landmarks creates a list of detected hands and their coordinates
        # I have used an if statemtent to ensure that if a hand is not shown in the frame, the empty results list counts as false logic and therefore skips over the line and does not crash the code
        if hand_landmarker_result.hand_landmarks:

            # I am creating a variable to store the results of the detected hand
            # The hand landmarker is set to 0 in the list because python counts from zero and there can only be one hand tracked at a time
            # There are 21 landmark points that can be tracked at once on one hand
            # I also had to change the round brackets to square brackets in this line becuase I am asking it to grab a number from a list, not run a function
            # If I am running it as a function, It doesn't send the results back to the list therefore the list is empty and the loop breaks, releasing the webcam
            hand = hand_landmarker_result.hand_landmarks[0]

            # This if statement check whether my switch is on or off, If SHOW_HANDS is false it will skip past this indented code and nothing will be drawn but the hand will still be tracked
            # If the switch is on or SHOW_HANDS is true it moves onto the code inside the indentation
            if SHOW_HAND:
              
                # Mediapipe's built in drawing tools help me draw the hands, meaning I dont have to convert landmark coordinates into pixels myself
                # Within my list of land mark drawings I have my bgr mp image, my hand list of 21 landmarks and the hand_connections list which has which landmarks are being drawn together
                mp_drawing.draw_landmarks(image, hand, list(HAND_CONNECTIONS))

            # Now, I have put these next lines of code out of the "If SHOW_HAND" loop so that my sound code runs even without the drawn landmarks
            # It is now back within the "if hand_landmarker_result.hand_landmarks" loop
            
            # I am going to refer back to the "x" variable before the loop, which represents the the last location detected hand from 0 (left screen) to 1 (right screen)
            # I want to control the sounds with the tip of my index finger, so I need to grab that specific landmark which is "hand[8]"
            # First, I need to limit my hand position x to a range of 0 to 1, so the sound stays within the edges of the screen and fade out
            # Inside the inner bracket, I have (hand[8].x, 0) which are the number being checked (hand location) and the lowest value allowed (0) on screen
            # "max" compares the two and picks the bigger number, so the result can never go below 0 
            # Inside the outer bracket, the two arguments are the result of the max and the highest value allowed (1) on screen
            # "min" picks the smaller of the two, so the result can never go above 1
            # Therefore I am creating a boundary which contains x into the range 0 to 1 just in case mp estimates a point slightly outside the frame
            x = min(max(hand[8].x, 0), 1)

            # Here I am increasing the value of the master sound by adding the value of the fade step
            # Using the same logic as before, I am using the "min" function to choose the smaller of the two values within the brackets, so master can never go above 1
            # Without this limit, master would keep climbing while my hand is on screen, and the fade out would be delayed
            master = min(master + FADE_STEP, 1)
            
            # I have added an else statement here so that when no hand is detected the master volume falls by the values of the FADE_STEP
            # x will always keep its last value, so by using the "max" function, the sound fades out slowly but never below 0
        else:
            master = max(master - FADE_STEP, 0)
            
            # These are the parameters that I have created for each sound from my hand position,
            # Rain is loudest at the left edge (0), wind in the middle (0.5) and birds at the right edge (1).
            # After testing the code I have altered the paramaters in hopes that the bird and rain sounds are better heard
            # Having their max volumes on the out most values of the screen 0 and 1 allowed them less room for their sounds to be played
            rain.set_volume(volume_for(x, 0.1, 0.4) * master)
            wind.set_volume(volume_for(x, 0.5, 0.25) * master)
            birds.set_volume(volume_for(x, 0.9, 0.4) * master)
            
        # This line displays the captured images in a separate pop up window that I can see on my screen and I am able to name the frame "Moving Through Sound"
        cv2.imshow('Moving Through Sound', image)
        
        # I have given the code a waitkey of 10 milliseconds, which means that between each frame that is being read the program waits for something to happen to my keyboard
        # If I press "q" on my keybpard during that time, the condition becomes true and breaks the loop
        # If I press another key or nothing at all, the 10 milliseconds expire and the condition becomes false and the loop begins again
        if cv2.waitKey(10) & 0xFF == ord('q'):
            break

# These lines are outside of the "with" statement block and they run once the loop has ended
# If the camera feed is broken or the frames cannot be read, I have told the code to release the video capture and delete all windows to prevent my script from crashing
# Instead it should move to the next line of code and "break" the loop to prevent major crashes and errors
# This turns off the webcam
cap.release()

# And this closes and removes all the display windows that opened during the code
cv2.destroyAllWindows()

# This stops all three sounds, so they do not keep playing after the code has finished.
pygame.mixer.stop()
```
